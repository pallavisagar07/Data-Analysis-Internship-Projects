## Task 2.1: Standardize Schedule Fields

The `Arrival_time` and `Departure_Time` columns were initially stored as object data types.

Since these columns contain time values in `HH:MM:SS` format, they were converted into Pandas datetime format using `pd.to_datetime()`.

This standardization is necessary for performing time-based calculations, particularly for calculating total journey duration in the next task.

```python
#Task 2.2 — Compute Total Journey Duration for Each Train

#Get the first and last schedule time for each train
journey_times = df.groupby('Train_No').agg(Start_Time=('Departure_Time', 'first'),
    End_Time=('Arrival_time', 'last')).reset_index()
journey_times
```

```python
#Calculate the journey duration
journey_times['Journey_Duration'] = (journey_times['End_Time'] - 
                    journey_times['Start_Time'])
journey_times.head()
```

```python
#check data type
journey_times['Journey_Duration'].dtype
```

```python
#Add 24 hours to overnight journeys
journey_times.loc[journey_times['Journey_Duration'] < pd.Timedelta(0),
    'Journey_Duration'] += pd.Timedelta(days=1)
journey_times.head()
```

```python
#Convert duration to hours
journey_times['Journey_Duration_Hours'] = (
    journey_times['Journey_Duration'].dt.total_seconds() / 3600)
journey_times.head()
```

```python
#Task 2.3 — Route Classification
df['Distance'].head()
```

```python
#Check the distance values
df[['Train_No', 'Station_Name', 'Distance']].head(20)
```

```python
df['Distance'].describe()
```

```python
#Calculate route distance for each train
route_distance = df.groupby('Train_No')['Distance'].max().reset_index()
```

```python
#Rename the column
route_distance = route_distance.rename(columns={'Distance': 'Route_Distance'})
route_distance.head(10)
```

```python
#Check the distribution
route_distance['Route_Distance'].describe()
```

```python
#Create the route categories
def classify_route(distance):
    if distance <= 38:
        return 'Short'
    elif distance <= 324:
        return 'Medium'
    else:
        return 'Long'
```

```python
#Apply the function
route_distance['Route_Category'] = route_distance['Route_Distance'].apply(
    classify_route)

route_distance.head(10)
```

```python
#Count each category
route_distance['Route_Category'].value_counts()
```

```python
#Calculate percentages
route_distance['Route_Category'].value_counts(normalize=True) * 100
```

```python
#Task 2.4 — Station-wise Train Frequency

#Calculate station frequency
station_frequency = df.groupby('Station_Name').size().reset_index(
    name='Train_Frequency')
station_frequency.head(10)             
```

```python
#Sort by frequency
station_frequency = station_frequency.sort_values(by='Train_Frequency',
    ascending=False)
station_frequency.head(10)           #View top 10 stations
```

# Level 2 — Train Schedule Analysis

## Objective

The objective of Level 2 was to perform further analysis on the train schedule data by standardizing time fields, calculating journey durations, classifying train routes based on distance, and analyzing station-wise train frequency.


### Task 2.1 — Standardize Schedule Time Fields

- Converted `Arrival_time` and `Departure_Time` into appropriate time/datetime values.
- Prepared the time fields for journey duration calculations.

### Task 2.2 — Calculate Journey Duration

- Identified the first departure time and last arrival time for each train.
- Calculated the total journey duration.
- Handled overnight journeys by accounting for midnight crossings.
- Converted journey duration into hours for easier analysis.

### Task 2.3 — Route Classification

- Calculated the total route distance for each train using the maximum cumulative distance.
- Classified routes into three categories:
  - **Short:** ≤ 38 km
  - **Medium:** > 38 km and ≤ 324 km
  - **Long:** > 324 km
- Quartile-based thresholds were used because the task did not specify fixed distance boundaries.

### Task 2.4 — Station-wise Train Frequency

- Calculated the number of unique trains serving each station.
- Identified the stations with the highest train frequency.
- Found a total of **8,099 unique stations**.

## Key Findings

- The dataset contains **11,113 unique trains**.

- Train routes were classified into Short, Medium, and Long categories based on route distance.
