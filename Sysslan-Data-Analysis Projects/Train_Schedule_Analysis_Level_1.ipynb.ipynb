```python
#Task1.2:List all trains with their starting and ending stations
df[['Train_No', 'Station_Code', 'Station_Name', 'Route_Number']].head(20)
```

```python
#Create a train route summary
train_routes = df.groupby('Train_No').agg(
    Starting_Station=('Station_Name', 'first'),
    Ending_Station=('Station_Name', 'last')).reset_index()
train_routes.head(10)
```

```python
#Include station codes too
train_routes = df.groupby('Train_No').agg(
    Starting_Station=('Station_Name', 'first'),
    Starting_Code=('Station_Code', 'first'),
    Ending_Station=('Station_Name', 'last'),
    Ending_Code=('Station_Code', 'last')).reset_index()
train_routes.head(10)
```

```python
#Check how many unique trains we have
print("Total unique trains:", df['Train_No'].nunique())
```

```python
#Task 1.3 — Calculate Number of Stops per Train
stops_per_train = df.groupby('Train_No').size().reset_index(
    name='Number_of_Stops')
stops_per_train
```

```python
# Top 10 trains with the most stops
stops_per_train.sort_values(by='Number_of_Stops',ascending=False).head(10)
```

```python
# Top 10 trains with the fewest stops
stops_per_train.sort_values(by='Number_of_Stops',ascending=True).head(10)
```

```python
#Task1.4: Identify trains with maximum and minimum stops

# maximum number of stops
max_stops = stops_per_train['Number_of_Stops'].max()
print("Maximum number of stops:", max_stops)
```

```python
#Find the trains having maximum stops
max_stop_trains = stops_per_train[
    stops_per_train['Number_of_Stops'] == max_stops]
max_stop_trains
```

```python
#Find the minimum number of stops
min_stops = stops_per_train['Number_of_Stops'].min()
print("Minimum number of stops:", min_stops)
```

```python
#Find the trains having minimum stops
min_stop_trains = stops_per_train[
    stops_per_train['Number_of_Stops'] == min_stops]
min_stop_trains
```
